a prediction and \"hidden state\" at each step, feeding its previous
hidden state into each next step. We take the final prediction to be the
output, i.e. which class the word belongs to.

of origin, and predict which language a name is from based on the
spelling.

Preparing the Data
==================

Download the data from
[here](https://download.pytorch.org/tutorial/data.zip) and extract it to
the current directory.

Included in the `data/names` directory are 18 text files named as
`[Language].txt`. Each file contains a bunch of names, one name per
line, mostly romanized (but we still need to convert from Unicode to
ASCII).

The first step is to define and clean our data. Initially, we need to
convert Unicode to plain ASCII to limit the RNN input layers. This is
accomplished by converting Unicode strings to ASCII and allowing only a
small set of allowed characters.


```python
#다양한 문자의 unicode를 컴퓨터가 처리할 수 있는 ASCII문자로 변환하는 과정(preprocessing)

import string #ascii문자 모듈
import unicodedata #unicode 모듈

# We can use "_" to represent an out-of-vocabulary character, that is, any character we are not handling in our model
allowed_characters = string.ascii_letters + " .,;'" + "_" #string.ascii_letters: a~z, A~Z 52자 / 특수문자 / _ =>사용 가능 문자
n_letters = len(allowed_characters) #허용된 문자 수

# Turn a Unicode string to plain ASCII, thanks to https://stackoverflow.com/a/518232/2809427
def unicodeToAscii(s): #유니코드->아스키 문자 변환 함수
    return ''.join(
        c for c in unicodedata.normalize('NFD', s)
        if unicodedata.category(c) != 'Mn'
        and c in allowed_characters
    ) #list comprehension
```

Here\'s an example of converting a unicode alphabet name to plain ASCII.
This simplifies the input layer


```python
print (f"converting 'Ślusàrski' to {unicodeToAscii('Ślusàrski')}")
```

Turning Names into Tensors
==========================

Now that we have all the names organized, we need to turn them into
Tensors to make any use of them.

To represent a single letter, we use a \"one-hot vector\" of size
`<1 x n_letters>`. A one-hot vector is filled with 0s except for a 1 at
index of the current letter, e.g. `"b" = <0 1 0 0 0 ...>`.

To make a word we join a bunch of those into a 2D matrix
`<line_length x 1 x n_letters>`.

That extra 1 dimension is because PyTorch assumes everything is in
batches - we\'re just using a batch size of 1 here.


```python
# Find letter index from all_letters, e.g. "a" = 0
#문자 -> ascii index 변환 함수
def letterToIndex(letter):
    # return our out-of-vocabulary character if we encounter a letter unknown to our model
    if letter not in allowed_characters:
        return allowed_characters.find("_") #금지 경우, "_" 반환
    else:
        return allowed_characters.find(letter) #해당 letter가 있는 index 반환

# Turn a line into a <line_length x 1(batch size) x n_letters> <-> 만약 한번에 5단어 처리시 => line_length x 5 x n_letters
# or an array of one-hot letter vectors => one-hot line vector 만듦 <0,0,...,0,1,0,...,0>
# one-hot line vector 만드는 함
def lineToTensor(line):
    tensor = torch.zeros(len(line), 1, n_letters)
    for li, letter in enumerate(line): #li: n번째 charachter, letter: 해당 문자의 ascii index -> 1(one-hot)
        tensor[li][0][letterToIndex(letter)] = 1
    return tensor #3D tensor 반환
```

Here are some examples of how to use `lineToTensor()` for a single and
multiple character string.


```python
print (f"The letter 'a' becomes {lineToTensor('a')}") #notice that the first position in the tensor = 1
print (f"The name 'Ahn' becomes {lineToTensor('Ahn')}") #depth:3, low:1, column: 52
```

Congratulations, you have built the foundational tensor objects for this
learning task! You can use a similar approach for other RNN tasks with
text.

Next, we need to combine all our examples into a dataset so we can
train, test and validate our models. For this, we will use the [Dataset
and
DataLoader](https://pytorch.org/tutorials/beginner/basics/data_tutorial.html)
classes to hold our dataset. Each Dataset needs to implement three
functions: `__init__`, `__len__`, and `__getitem__`.


```python
from io import open #파일 오픈 함수
import glob #특정 패턴(.txt)가진 파일 찾도록 도움
import os #운영체제 모듈
import time

import torch
from torch.utils.data import Dataset #PyTorch에서 class 만들 때, 무조건 상속받아야 하는 부모 클래스

class NamesDataset(Dataset):

    def __init__(self, data_dir):
        self.data_dir = data_dir #data_dir: 파일 경로 input param으로 받음, 파일 경로 기록
        self.load_time = time.localtime #파일 로드 시간 기록
        labels_set = set() #공백 세트: 중복 이름 배제(ex.{"kim","ahn",...,"kim"(x)})

        self.data = [] #이름 문자열(kim)
        self.data_tensors = [] #이름->텐서
        self.labels = [] #레이블 문자열(korea)
        self.labels_tensors = [] #레이블->텐서

        #read all the ``.txt`` files in the specified directory
        text_files = glob.glob(os.path.join(data_dir, '*.txt')) #입력 받은 파일 경로+*txt파일 합침->(~*.txt형태)->모든 해당 경로 txt파일 리스트 반환
        for filename in text_files:
            label = os.path.splitext(os.path.basename(filename))[0] #파일명만 가져옴:basename(filename): ~/data/korean.txt -> korea.txt // splitext[korea.txt] -> [korea, txt]
            labels_set.add(label)
            lines = open(filename, encoding='utf-8').read().strip().split('\n') #file 내부 전체 가져옴 -> 널 문자 제거 -> 개행문자(\n)경계로 리스트 집
            for name in lines:
                self.data.append(name)
                self.data_tensors.append(lineToTensor(name))
                self.labels.append(label)

        #Cache the tensor representation of the labels
        self.labels_uniq = list(labels_set) #label set -> list 변환: 순서 고정 위하여
        for idx in range(len(self.labels)):
            temp_tensor = torch.tensor([self.labels_uniq.index(self.labels[idx])], dtype=torch.long) #현재 label이 list에서 몇번째 순서인지 찾음(ex, korea=2 ->[2])
            self.labels_tensors.append(temp_tensor) #찾은 순서 저장([2])

    def __len__(self):
        return len(self.data) #데이터(이름) 개수 반환

    def __getitem__(self, idx): #index 반환 함수 호출 시, 실행됨
        data_item = self.data[idx]
        data_label = self.labels[idx]
        data_tensor = self.data_tensors[idx]
        label_tensor = self.labels_tensors[idx]

        return label_tensor, data_tensor, data_label, data_item
```

Here we can load our example data into the `NamesDataset`


```python
alldata = NamesDataset("data/names")
print(f"loaded {len(alldata)} items of data")
print(f"example = {alldata[5]}")
```

Creating the Network
====================

Before autograd, creating a recurrent neural network in Torch involved
cloning the parameters of a layer over several timesteps. The layers
held hidden state and gradients which are now entirely handled by the
graph itself. This means you can implement a RNN in a very \"pure\" way,
as regular feed-forward layers.

This CharRNN class implements an RNN with three components. First, we
use the [nn.RNN
implementation](https://pytorch.org/docs/stable/generated/torch.nn.RNN.html).
Next, we define a layer that maps the RNN hidden layers to our output.
And finally, we apply a `softmax` function. Using `nn.RNN` leads to a
significant improvement in performance, such as cuDNN-accelerated
kernels, versus implementing each layer as a `nn.Linear`. It also
simplifies the implementation in `forward()`.


```python
import torch.nn as nn
import torch.nn.functional as F

class CharRNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size): #input size: n_letters(size of one hot vec)들어감 = 글자 하나씩 들어가는 것
        super(CharRNN, self).__init__()

        self.rnn = nn.RNN(input_size, hidden_size) #rnn layers
        self.h2o = nn.Linear(hidden_size, output_size) #output layer(linear projection)
        self.softmax = nn.LogSoftmax(dim=1) #prediction위해 softmax, dim=0(batch), 1(class)

    def forward(self, line_tensor): #line_tensor: 이름 전체 3d tensor
        rnn_out, hidden = self.rnn(line_tensor) #rnn_out: t sequence에 따른 결과 모두 저장 / hidden: Last hidden state(=h_T)
        #hidden = (Number of layers, Batch size, Hidden size) 1층 RNN 사용함 -> hidden[0] 벗기면 (batch size, hidden size)모양 나옴 -> output size 반
        output = self.h2o(hidden[0])
        output = self.softmax(output)

        return output
```

We can then create an RNN with 58 input nodes, 128 hidden nodes, and 18
outputs:


```python
n_hidden = 128
rnn = CharRNN(n_letters, n_hidden, len(alldata.labels_uniq))
print(rnn)
```

After that we can pass our Tensor to the RNN to obtain a predicted
output. Subsequently, we use a helper function, `label_from_output`, to
derive a text label for the class.


```python
def label_from_output(output, output_labels):
    top_n, top_i = output.topk(1) #top_n:output의 최댓값, top_i:최댓값의 index
    label_i = top_i[0].item() #top_i는 2D tensor->첫번째 요소 꺼내고 해당 index값 벗김: tensor->integer 변환
    return output_labels[label_i], label_i

input = lineToTensor('kim')
output = rnn(input) #this is equivalent to ``output = rnn.forward(input)``
print(output)
print(label_from_output(output, alldata.labels_uniq))
```

Training the Network
====================

Now all it takes to train this network is show it a bunch of examples,
have it make guesses, and tell it if it\'s wrong.

We do this by defining a `train()` function which trains the model on a
given dataset using minibatches. RNNs RNNs are trained similarly to
other networks; therefore, for completeness, we include a batched
training method here. The loop (`for i in batch`) computes the losses
for each of the items in the batch before adjusting the weights. This
operation is repeated until the number of epochs is reached.


```python
import random
import numpy as np

#training function 정의
def train(rnn, training_data, n_epoch = 10, n_batch_size = 64, report_every = 50, learning_rate = 0.2, criterion = nn.NLLLoss()): #criterion(채점기준):negative loglikelihood
    """
    Learn on a batch of training_data for a specified number of iterations and reporting thresholds
    """
    # Keep track of losses for plotting
    current_loss = 0
    all_losses = []
    rnn.train()
    optimizer = torch.optim.SGD(rnn.parameters(), lr=learning_rate) #Stochastic Gradient Decent

    start = time.time()
    print(f"training on data set with n = {len(training_data)}")

    for iter in range(1, n_epoch + 1):
        rnn.zero_grad() # clear the gradients

        # create some minibatches
        # we cannot use dataloaders because each of our names is a different length
        batches = list(range(len(training_data)))
        random.shuffle(batches)
        batches = np.array_split(batches, len(batches) //n_batch_size ) # batches 데이터(0...17062 셔플)를 batch_size에 알맞게 나눔(split)

        for idx, batch in enumerate(batches):
            batch_loss = 0
            for i in batch: #for each example in this batch
                (label_tensor, text_tensor, label, text) = training_data[i]
                output = rnn.forward(text_tensor)
                loss = criterion(output, label_tensor) #Negative LogLikelihood Loss function
                batch_loss += loss

            # optimize parameters
            batch_loss.backward() #backpropagaion: 미분값 저장
            nn.utils.clip_grad_norm_(rnn.parameters(), 3) #grad cliping: 기울기가 3을 넘어가면 강제로 3으로 줄임
            optimizer.step() #SGD 수행
            optimizer.zero_grad() #optimzie 이후 gradient 초기화(다음 수행을 위함)

            current_loss += batch_loss.item() / len(batch)

        all_losses.append(current_loss / len(batches) )
        if iter % report_every == 0:
            print(f"{iter} ({iter / n_epoch:.0%}): \t average batch loss = {all_losses[-1]}") #all_losses[-1]: 가장 최신 loss
        current_loss = 0

    return all_losses #오차 기록 반환
```

We can now train a dataset with minibatches for a specified number of
epochs. The number of epochs for this example is reduced to speed up the
build. You can get better results with different parameters.


```python
start = time.time()
all_losses = train(rnn, train_set, n_epoch=27, learning_rate=0.15, report_every=5)
end = time.time()
print(f"training took {end-start}s")
```

Evaluating the Results
======================

To see how well the network performs on different categories, we will
create a confusion matrix, indicating for every actual language (rows)
which language the network guesses (columns). To calculate the confusion
matrix a bunch of samples are run through the network with `evaluate()`,
which is the same as `train()` minus the backprop.


```python
#Test set의 정확성 보기 위함
def evaluate(rnn, testing_data, classes):
    confusion = torch.zeros(len(classes), len(classes)) #[N,N] 각 class에 대한 matrix

    rnn.eval() #set to eval mode: dropout, normalization이 training때와 다르게 동작
    with torch.no_grad(): # do not record the gradients during eval phase
        for i in range(len(testing_data)):
            (label_tensor, text_tensor, label, text) = testing_data[i]
            output = rnn(text_tensor)
            guess, guess_i = label_from_output(output, classes) #예측 이름, 해당 index(ex. 'japanese', 5 출력)
            label_i = classes.index(label) #실제 정답의 index 뽑음
            confusion[label_i][guess_i] += 1 #상관관계 matrix 제작

    # Normalize by dividing every row by its sum
    for i in range(len(classes)):
        denom = confusion[i].sum() #row의 합
        if denom > 0:
            confusion[i] = confusion[i] / denom #row의 모든 element를 sum of row로 나눔 -> normalization: total sum = 1

    # Set up plot
    fig = plt.figure()
    ax = fig.add_subplot(111)
    cax = ax.matshow(confusion.cpu().numpy()) #numpy uses cpu here so we need to use a cpu version: tensor -> numpy로 변환
    fig.colorbar(cax)

    # Set up axes
    ax.set_xticks(np.arange(len(classes)), labels=classes, rotation=90)
    ax.set_yticks(np.arange(len(classes)), labels=classes)

    # Force label at every tick
    ax.xaxis.set_major_locator(ticker.MultipleLocator(1)) #ticker.MultipleLocator:눈금 한 칸마다 그리도록
    ax.yaxis.set_major_locator(ticker.MultipleLocator(1))

    # sphinx_gallery_thumbnail_number = 2
    plt.show()



evaluate(rnn, test_set, classes=alldata.labels_uniq)
```

You can pick out bright spots off the main axis that show which
languages it guesses incorrectly, e.g. Chinese for Korean, and Spanish
for Italian. It seems to do very well with Greek, and very poorly with
English (perhaps because of overlap with other languages).
